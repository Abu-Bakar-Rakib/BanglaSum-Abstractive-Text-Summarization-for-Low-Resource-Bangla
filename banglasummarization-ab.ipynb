```python
# ============================================================================
# 2. Imports & Configuration
# ============================================================================
import os
import json
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from tqdm.auto import tqdm
from pathlib import Path

import torch
from torch.utils.data import Dataset, DataLoader

from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
    Seq2SeqTrainer,
    Seq2SeqTrainingArguments,
    DataCollatorForSeq2Seq,
    EarlyStoppingCallback,
    set_seed,
)
from rouge_score import rouge_scorer
from bert_score import score as bert_score_fn
import nltk
nltk.download('punkt', quiet=True)
nltk.download('wordnet', quiet=True)

warnings.filterwarnings('ignore')

# ── Reproducibility ──────────────────────────────────────────────────────────
SEED = 42
set_seed(SEED)
random.seed(SEED)
np.random.seed(SEED)

# ── Paths ─────────────────────────────────────────────────────────────────────
OUTPUT_DIR     = Path("./banglasum_output")
RESULTS_DIR    = Path("./banglasum_results")
FIGURES_DIR    = Path("./banglasum_figures")
for d in [OUTPUT_DIR, RESULTS_DIR, FIGURES_DIR]:
    d.mkdir(exist_ok=True)

# ── Config ────────────────────────────────────────────────────────────────────
MODEL_NAME      = "csebuetnlp/mT5_multilingual_XLSum"
MAX_INPUT_LEN   = 512
MAX_TARGET_LEN  = 128
TRAIN_BATCH     = 1
EVAL_BATCH      = 2
GRAD_ACCUM      = 8          # effective batch = 8
LR              = 5e-5
NUM_EPOCHS      = 3
WARMUP_STEPS    = 200
NUM_BEAMS       = 4
WHITESPACE_HANDLER = lambda k: ' '.join(k.strip().split())

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"✅ Device : {DEVICE}")
print(f"✅ Model  : {MODEL_NAME}")
print(f"✅ Outputs: {OUTPUT_DIR}")


# ============================================================================
# 3. Plotting Theme Setup
# ============================================================================
# ── Custom dark-teal palette inspired by Bangla flag colors ──────────────────
PALETTE = {
    'primary'   : '#006A4E',   # deep Bangla green
    'secondary' : '#F42A41',   # Bangla red
    'accent'    : '#FFB703',   # gold
    'blue'      : '#219EBC',
    'light'     : '#8ECAE6',
    'bg'        : '#F8F9FA',
    'dark'      : '#212529',
}
COLORS = [PALETTE['primary'], PALETTE['secondary'],
          PALETTE['accent'],  PALETTE['blue'], PALETTE['light']]

plt.rcParams.update({
    'figure.facecolor'  : PALETTE['bg'],
    'axes.facecolor'    : 'white',
    'axes.edgecolor'    : '#DEE2E6',
    'axes.grid'         : True,
    'grid.alpha'        : 0.4,
    'grid.linestyle'    : '--',
    'font.family'       : 'DejaVu Sans',
    'font.size'         : 11,
    'axes.titlesize'    : 13,
    'axes.titleweight'  : 'bold',
    'axes.labelsize'    : 11,
    'xtick.labelsize'   : 10,
    'ytick.labelsize'   : 10,
    'legend.frameon'    : False,
    'savefig.dpi'       : 150,
    'savefig.bbox'      : 'tight',
})

def save_fig(name):
    plt.savefig(FIGURES_DIR / f"{name}.png", dpi=150, bbox_inches='tight')
    plt.show()
    plt.close()

print("✅ Plot theme ready")


# ============================================================================
# 4. Load & Explore Dataset
# ============================================================================
def load_and_explore_dataset():
    print("⏳ Loading XL-Sum (Bengali) ...")
    raw = load_dataset("csebuetnlp/xlsum", "bengali", trust_remote_code=True)
    print(raw)

    train_df = raw['train'].to_pandas()
    val_df   = raw['validation'].to_pandas()
    test_df  = raw['test'].to_pandas()

    print(f"\n📊 Split sizes")
    print(f"   Train : {len(train_df):,}")
    print(f"   Val   : {len(val_df):,}")
    print(f"   Test  : {len(test_df):,}")

    # ── Length distributions ─────────────────────────────────────────────────
    train_df['article_len'] = train_df['text'].apply(lambda x: len(x.split()))
    train_df['summary_len'] = train_df['summary'].apply(lambda x: len(x.split()))
    train_df['compression'] = train_df['summary_len'] / train_df['article_len']

    fig, axes = plt.subplots(1, 3, figsize=(16, 4))
    fig.suptitle('XL-Sum Bengali — Dataset Statistics', fontsize=15, fontweight='bold', y=1.02)

    # Article length
    axes[0].hist(train_df['article_len'].clip(upper=800), bins=40,
                 color=PALETTE['primary'], edgecolor='white', alpha=0.85)
    axes[0].set_title('Article Word Count')
    axes[0].set_xlabel('Words')
    axes[0].set_ylabel('Frequency')
    axes[0].axvline(train_df['article_len'].median(), color=PALETTE['secondary'],
                    linestyle='--', linewidth=1.5, label=f"Median: {train_df['article_len'].median():.0f}")
    axes[0].legend()

    # Summary length
    axes[1].hist(train_df['summary_len'].clip(upper=150), bins=40,
                 color=PALETTE['blue'], edgecolor='white', alpha=0.85)
    axes[1].set_title('Summary Word Count')
    axes[1].set_xlabel('Words')
    axes[1].axvline(train_df['summary_len'].median(), color=PALETTE['secondary'],
                    linestyle='--', linewidth=1.5, label=f"Median: {train_df['summary_len'].median():.0f}")
    axes[1].legend()

    # Compression ratio
    axes[2].hist(train_df['compression'].clip(upper=0.5), bins=40,
                 color=PALETTE['accent'], edgecolor='white', alpha=0.85)
    axes[2].set_title('Compression Ratio (summary/article)')
    axes[2].set_xlabel('Ratio')
    axes[2].axvline(train_df['compression'].median(), color=PALETTE['secondary'],
                    linestyle='--', linewidth=1.5, label=f"Median: {train_df['compression'].median():.3f}")
    axes[2].legend()

    plt.tight_layout()
    save_fig('01_dataset_statistics')

    # Stats table
    stats = train_df[['article_len','summary_len','compression']].describe().round(2)
    print(stats)

    # ── Sample inspection ────────────────────────────────────────────────────
    sample = train_df.sample(1, random_state=SEED).iloc[0]
    print("━" * 70)
    print("📰 ARTICLE (first 300 chars):")
    print(sample['text'][:300], "...")
    print("\n📝 REFERENCE SUMMARY:")
    print(sample['summary'])
    print("━" * 70)

    return raw, train_df, val_df, test_df


# ============================================================================
# 5. Tokenizer & Preprocessing
# ============================================================================
def build_tokenizer_and_tokenize(raw):
    print("⏳ Loading tokenizer ...")
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

    def preprocess(examples):
        articles  = [WHITESPACE_HANDLER(a) for a in examples['text']]
        summaries = [WHITESPACE_HANDLER(s) for s in examples['summary']]

        model_inputs = tokenizer(
            articles,
            max_length=MAX_INPUT_LEN,
            truncation=True,
            padding=False,
        )
        labels = tokenizer(
            text_target=summaries,
            max_length=MAX_TARGET_LEN,
            truncation=True,
            padding=False,
        )
        model_inputs['labels'] = labels['input_ids']
        return model_inputs

    tokenized = raw.map(
        preprocess,
        batched=True,
        remove_columns=raw['train'].column_names,
        desc="Tokenizing",
    )
    print("✅ Tokenization complete")
    print(tokenized)

    return tokenizer, tokenized


# ============================================================================
# 6. Model Loading & Fine-Tuning
# ============================================================================
def load_model():
    print("⏳ Loading model ...")
    import gc
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        gc.collect()
        
    model = AutoModelForSeq2SeqLM.from_pretrained(
        MODEL_NAME,
        low_cpu_mem_usage=True,
    )
    model = model.to(DEVICE)

    # Freeze the massive multilingual embeddings to save VRAM (from 966M down to ~300M trainable params)
    for name, param in model.named_parameters():
        if "embed" in name or "shared" in name or "lm_head" in name:
            param.requires_grad = False

    total_params     = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"✅ Total params    : {total_params/1e6:.1f}M")
    print(f"✅ Trainable params: {trainable_params/1e6:.1f}M")

    return model


def build_trainer(model, tokenizer, tokenized):
    data_collator = DataCollatorForSeq2Seq(
        tokenizer,
        model=model,
        label_pad_token_id=-100,
        pad_to_multiple_of=8,
    )

    training_args = Seq2SeqTrainingArguments(
        output_dir                  = str(OUTPUT_DIR),
        num_train_epochs            = NUM_EPOCHS,
        per_device_train_batch_size = TRAIN_BATCH,
        per_device_eval_batch_size  = EVAL_BATCH,
        gradient_accumulation_steps = GRAD_ACCUM,
        learning_rate               = LR,
        warmup_steps                = WARMUP_STEPS,
        weight_decay                = 0.01,
        optim                       = 'adafactor',
        fp16                        = (DEVICE.type == 'cuda'),
        gradient_checkpointing      = True,
        predict_with_generate       = True,
        generation_max_length       = MAX_TARGET_LEN,
        generation_num_beams        = NUM_BEAMS,
        eval_strategy               = "epoch",
        save_strategy               = "epoch",
        load_best_model_at_end      = True,
        metric_for_best_model       = "eval_loss",
        logging_dir                 = str(OUTPUT_DIR / 'logs'),
        logging_steps               = 50,
        report_to                   = 'none',
        seed                        = SEED,
    )

    trainer = Seq2SeqTrainer(
        model           = model,
        args            = training_args,
        train_dataset   = tokenized['train'],
        eval_dataset    = tokenized['validation'],
        processing_class= tokenizer,
        data_collator   = data_collator,
        callbacks       = [EarlyStoppingCallback(early_stopping_patience=2)],
    )

    print("✅ Trainer ready")
    return trainer


def train_model(trainer, tokenizer):
    print("🚀 Starting fine-tuning ...")
    train_result = trainer.train()
    print("\n✅ Training complete!")
    print(train_result.metrics)

    # Save model & tokenizer
    trainer.save_model(str(OUTPUT_DIR / 'best_model'))
    tokenizer.save_pretrained(str(OUTPUT_DIR / 'best_model'))
    print(f"✅ Model saved to {OUTPUT_DIR / 'best_model'}")

    return train_result


# ============================================================================
# 7. Training Curves
# ============================================================================
def plot_training_curves(trainer):
    log_history = trainer.state.log_history

    train_logs = [(l['epoch'], l['loss']) for l in log_history if 'loss' in l and 'eval_loss' not in l]
    eval_logs  = [(l['epoch'], l['eval_loss']) for l in log_history if 'eval_loss' in l]

    train_epochs, train_losses = zip(*train_logs) if train_logs else ([], [])
    eval_epochs,  eval_losses  = zip(*eval_logs)  if eval_logs  else ([], [])

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(train_epochs, train_losses, color=PALETTE['primary'], linewidth=2,
            label='Train Loss', alpha=0.9)
    ax.plot(eval_epochs,  eval_losses,  color=PALETTE['secondary'], linewidth=2,
            linestyle='--', marker='o', markersize=6, label='Validation Loss')
    ax.set_title('BanglaSum — Training & Validation Loss')
    ax.set_xlabel('Epoch')
    ax.set_ylabel('Cross-Entropy Loss')
    ax.legend()
    ax.fill_between(train_epochs, train_losses, alpha=0.08, color=PALETTE['primary'])
    plt.tight_layout()
    save_fig('02_training_curves')


# ============================================================================
# 8. Generate Summaries on Test Set
# ============================================================================
def generate_summary(article, model, tokenizer, device,
                     max_input=MAX_INPUT_LEN, max_new_tokens=MAX_TARGET_LEN,
                     num_beams=NUM_BEAMS):
    inputs = tokenizer(
        [WHITESPACE_HANDLER(article)],
        return_tensors='pt',
        padding=True,
        truncation=True,
        max_length=max_input,
    ).to(device)

    with torch.no_grad():
        ids = model.generate(
            **inputs,
            num_beams=num_beams,
            max_new_tokens=max_new_tokens,
            no_repeat_ngram_size=2,
            early_stopping=True,
        )
    return tokenizer.decode(ids[0], skip_special_tokens=True,
                            clean_up_tokenization_spaces=True)


def generate_on_test_set(model, tokenizer, test_df):
    print("⏳ Generating summaries on test set (first 200 samples) ...")
    model.eval()

    N_TEST          = min(200, len(test_df))
    test_sample     = test_df.sample(N_TEST, random_state=SEED).reset_index(drop=True)
    generated_sums  = []

    for _, row in tqdm(test_sample.iterrows(), total=N_TEST):
        gen = generate_summary(row['text'], model, tokenizer, DEVICE)
        generated_sums.append(gen)

    test_sample['generated_summary'] = generated_sums
    print("✅ Generation complete")

    return test_sample, N_TEST


# ============================================================================
# 9. Automatic Evaluation Metrics
# ============================================================================
def compute_rouge(test_sample):
    # ── ROUGE ────────────────────────────────────────────────────────────────
    # Monkey-patch the default rouge tokenizer to support Bengali 
    # (otherwise it strips all non-English characters, leading to 0.0 scores)
    import rouge_score.tokenize as rouge_tokenize
    original_tokenize = rouge_tokenize.tokenize
    rouge_tokenize.tokenize = lambda text, stemmer: text.strip().split()

    scorer = rouge_scorer.RougeScorer(['rouge1','rouge2','rougeL'], use_stemmer=False)

    r1_list, r2_list, rl_list = [], [], []
    for ref, hyp in zip(test_sample['summary'], test_sample['generated_summary']):
        s = scorer.score(ref, hyp)
        r1_list.append(s['rouge1'].fmeasure)
        r2_list.append(s['rouge2'].fmeasure)
        rl_list.append(s['rougeL'].fmeasure)

    rouge_scores = {
        'ROUGE-1': np.mean(r1_list),
        'ROUGE-2': np.mean(r2_list),
        'ROUGE-L': np.mean(rl_list),
    }

    test_sample['rouge1'] = r1_list
    test_sample['rouge2'] = r2_list
    test_sample['rougeL'] = rl_list

    rouge_tokenize.tokenize = original_tokenize

    print("ROUGE Scores:")
    for k, v in rouge_scores.items():
        print(f"  {k}: {v:.4f}")

    return scorer, rouge_scores, r1_list, r2_list, rl_list


def compute_bertscore(test_sample):
    # ── BERTScore ─────────────────────────────────────────────────────────────
    print("⏳ Computing BERTScore (this may take a few minutes) ...")
    P, R, F1 = bert_score_fn(
        cands=test_sample['generated_summary'].tolist(),
        refs=test_sample['summary'].tolist(),
        lang='bn',
        model_type='bert-base-multilingual-cased',
        verbose=False,
    )
    bert_f1_mean = F1.mean().item()
    test_sample['bertscore_f1'] = F1.numpy()
    print(f"  BERTScore F1: {bert_f1_mean:.4f}")

    return F1, bert_f1_mean


def summarize_metrics(test_sample, rouge_scores, bert_f1_mean, r1_list, r2_list, rl_list, F1):
    # ── Summary length analysis ───────────────────────────────────────────────
    test_sample['gen_len'] = test_sample['generated_summary'].apply(lambda x: len(x.split()))
    test_sample['ref_len'] = test_sample['summary'].apply(lambda x: len(x.split()))

    # ── Consolidated metrics table ────────────────────────────────────────────
    metrics_summary = {
        'Metric'   : ['ROUGE-1', 'ROUGE-2', 'ROUGE-L', 'BERTScore F1'],
        'Score'    : [rouge_scores['ROUGE-1'], rouge_scores['ROUGE-2'],
                      rouge_scores['ROUGE-L'], bert_f1_mean],
        'Std'      : [np.std(r1_list), np.std(r2_list),
                      np.std(rl_list), F1.numpy().std()],
    }
    metrics_df = pd.DataFrame(metrics_summary)
    metrics_df['Score_%'] = (metrics_df['Score'] * 100).round(2)
    metrics_df['Std_%']   = (metrics_df['Std']   * 100).round(2)
    print("\n📊 Evaluation Metrics Summary:")
    print(metrics_df.to_string(index=False))

    return metrics_df


# ============================================================================
# 10. Evaluation Visualizations
# ============================================================================
def plot_metric_scores(metrics_df):
    # ── (A) Bar chart: Metric scores ──────────────────────────────────────────
    fig, ax = plt.subplots(figsize=(9, 5))
    bars = ax.barh(
        metrics_df['Metric'],
        metrics_df['Score_%'],
        xerr=metrics_df['Std_%'],
        color=COLORS[:4],
        edgecolor='white',
        height=0.55,
        capsize=4,
        error_kw={'linewidth': 1.5},
    )
    for bar, val in zip(bars, metrics_df['Score_%']):
        ax.text(val + 0.5, bar.get_y() + bar.get_height()/2,
                f'{val:.2f}', va='center', fontsize=11, fontweight='bold')
    ax.set_xlim(0, 100)
    ax.set_xlabel('Score (%)')
    ax.set_title('BanglaSum — Automatic Evaluation Metrics')
    ax.invert_yaxis()
    plt.tight_layout()
    save_fig('03_metric_scores')


def plot_rouge_distributions(test_sample):
    # ── (B) Distribution of per-sample ROUGE scores ───────────────────────────
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    fig.suptitle('Per-Sample ROUGE Score Distributions', fontsize=14, fontweight='bold')

    for ax, col, color, label in zip(
        axes,
        ['rouge1','rouge2','rougeL'],
        [PALETTE['primary'], PALETTE['blue'], PALETTE['accent']],
        ['ROUGE-1','ROUGE-2','ROUGE-L'],
    ):
        data = test_sample[col]
        ax.hist(data, bins=30, color=color, edgecolor='white', alpha=0.85)
        ax.axvline(data.mean(), color=PALETTE['secondary'], linestyle='--',
                   linewidth=2, label=f'Mean: {data.mean():.3f}')
        ax.axvline(data.median(), color=PALETTE['dark'], linestyle=':',
                   linewidth=1.5, label=f'Median: {data.median():.3f}')
        ax.set_title(label)
        ax.set_xlabel('Score')
        ax.set_ylabel('Count')
        ax.legend(fontsize=9)

    plt.tight_layout()
    save_fig('04_rouge_distributions')


def plot_rouge_vs_bertscore(test_sample):
    # ── (C) Scatter: ROUGE-1 vs BERTScore ────────────────────────────────────
    fig, ax = plt.subplots(figsize=(8, 6))
    sc = ax.scatter(
        test_sample['rouge1'],
        test_sample['bertscore_f1'],
        c=test_sample['rouge2'],
        cmap='YlGn',
        alpha=0.6,
        edgecolors='none',
        s=40,
    )
    cb = plt.colorbar(sc, ax=ax)
    cb.set_label('ROUGE-2', fontsize=10)
    ax.set_xlabel('ROUGE-1')
    ax.set_ylabel('BERTScore F1')
    ax.set_title('ROUGE-1 vs BERTScore F1\n(colored by ROUGE-2)')

    # Correlation
    corr = np.corrcoef(test_sample['rouge1'], test_sample['bertscore_f1'])[0,1]
    ax.text(0.05, 0.95, f'Pearson r = {corr:.3f}', transform=ax.transAxes,
            fontsize=11, va='top',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    plt.tight_layout()
    save_fig('05_rouge_vs_bertscore')


def plot_length_analysis(test_sample):
    # ── (D) Generated vs Reference length comparison ──────────────────────────
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle('Summary Length Analysis', fontsize=14, fontweight='bold')

    # Overlapping histograms
    axes[0].hist(test_sample['ref_len'], bins=30, alpha=0.65,
                 color=PALETTE['primary'], label='Reference', edgecolor='white')
    axes[0].hist(test_sample['gen_len'], bins=30, alpha=0.65,
                 color=PALETTE['secondary'], label='Generated', edgecolor='white')
    axes[0].set_xlabel('Word Count')
    axes[0].set_ylabel('Frequency')
    axes[0].set_title('Length Distribution')
    axes[0].legend()

    # Scatter: ref len vs gen len
    axes[1].scatter(test_sample['ref_len'], test_sample['gen_len'],
                    alpha=0.4, color=PALETTE['blue'], s=25)
    lim = max(test_sample['ref_len'].max(), test_sample['gen_len'].max()) + 5
    axes[1].plot([0, lim], [0, lim], 'k--', linewidth=1, alpha=0.5, label='y = x')
    axes[1].set_xlabel('Reference Length (words)')
    axes[1].set_ylabel('Generated Length (words)')
    axes[1].set_title('Reference vs Generated Length')
    axes[1].legend()

    plt.tight_layout()
    save_fig('06_length_analysis')


def plot_metric_boxplots(test_sample):
    # ── (E) Box plots of all metrics ──────────────────────────────────────────
    fig, ax = plt.subplots(figsize=(9, 5))
    metric_cols = ['rouge1','rouge2','rougeL','bertscore_f1']
    metric_labels = ['ROUGE-1','ROUGE-2','ROUGE-L','BERTScore F1']

    bp = ax.boxplot(
        [test_sample[c] for c in metric_cols],
        labels=metric_labels,
        patch_artist=True,
        medianprops=dict(color=PALETTE['secondary'], linewidth=2),
        whiskerprops=dict(linewidth=1.5),
        capprops=dict(linewidth=1.5),
        flierprops=dict(marker='o', markersize=3, alpha=0.4),
    )
    for patch, color in zip(bp['boxes'], COLORS):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    ax.set_ylabel('Score')
    ax.set_title('Score Distribution Across Metrics')
    plt.tight_layout()
    save_fig('07_metric_boxplots')


# ============================================================================
# 11. Qualitative Analysis — Best / Worst Examples
# ============================================================================
def display_example(row, rank_label):
    print(f"{'━'*70}")
    print(f"🏷️  {rank_label}  |  ROUGE-1: {row['rouge1']:.3f}  |  BERTScore: {row['bertscore_f1']:.3f}")
    print(f"\n📰 ARTICLE (truncated to 200 chars):")
    print(row['text'][:200] + " ...")
    print(f"\n✅ REFERENCE SUMMARY:")
    print(row['summary'])
    print(f"\n🤖 GENERATED SUMMARY:")
    print(row['generated_summary'])
    print()


def qualitative_analysis(test_sample):
    # Top-3 best
    print("\n" + "═"*70)
    print("🌟 TOP 3 BEST SUMMARIES (by ROUGE-1)")
    print("═"*70)
    for i, (_, row) in enumerate(test_sample.nlargest(3, 'rouge1').iterrows()):
        display_example(row, f"BEST #{i+1}")

    # Top-3 worst
    print("\n" + "═"*70)
    print("⚠️  TOP 3 WORST SUMMARIES (by ROUGE-1)")
    print("═"*70)
    for i, (_, row) in enumerate(test_sample.nsmallest(3, 'rouge1').iterrows()):
        display_example(row, f"WORST #{i+1}")


# ============================================================================
# 12. Baseline Comparison
# ============================================================================
def textrank_summary(text, sentences=2):
    from sumy.parsers.plaintext import PlaintextParser
    from sumy.nlp.tokenizers import Tokenizer
    from sumy.summarizers.text_rank import TextRankSummarizer
    try:
        parser = PlaintextParser.from_string(text, Tokenizer('english'))
        summarizer = TextRankSummarizer()
        sents = summarizer(parser.document, sentences)
        return ' '.join(str(s) for s in sents)
    except:
        return text[:100]


def baseline_comparison(test_sample, scorer):
    # ── TextRank extractive baseline (sumy) ──────────────────────────────────
    # Run once in your shell:  pip install sumy -q

    print("⏳ Computing TextRank baseline ...")
    # Use a small sample for speed
    baseline_sample = test_sample.head(50).copy()
    baseline_sample['textrank_summary'] = baseline_sample['text'].apply(textrank_summary)

    # ROUGE for TextRank
    tr_r1, tr_r2, tr_rl = [], [], []
    for ref, hyp in zip(baseline_sample['summary'], baseline_sample['textrank_summary']):
        s = scorer.score(ref, hyp)
        tr_r1.append(s['rouge1'].fmeasure)
        tr_r2.append(s['rouge2'].fmeasure)
        tr_rl.append(s['rougeL'].fmeasure)

    # BanglaBART on same 50 samples
    mt5_r1 = test_sample.head(50)['rouge1'].tolist()
    mt5_r2 = test_sample.head(50)['rouge2'].tolist()
    mt5_rl = test_sample.head(50)['rougeL'].tolist()

    comparison = {
        'Model'  : ['TextRank (Extractive)', 'BanglaBART Fine-tuned (Ours)'],
        'ROUGE-1': [np.mean(tr_r1)*100, np.mean(mt5_r1)*100],
        'ROUGE-2': [np.mean(tr_r2)*100, np.mean(mt5_r2)*100],
        'ROUGE-L': [np.mean(tr_rl)*100, np.mean(mt5_rl)*100],
    }
    comp_df = pd.DataFrame(comparison).round(2)
    print("\n📊 Model Comparison:")
    print(comp_df.to_string(index=False))

    # ── Grouped bar: Model comparison ────────────────────────────────────────
    fig, ax = plt.subplots(figsize=(10, 5))
    x = np.arange(3)
    width = 0.35

    b1 = ax.bar(x - width/2, comp_df.iloc[0][['ROUGE-1','ROUGE-2','ROUGE-L']],
                width, label='TextRank (Extractive)',
                color=PALETTE['light'], edgecolor='white')
    b2 = ax.bar(x + width/2, comp_df.iloc[1][['ROUGE-1','ROUGE-2','ROUGE-L']],
                width, label='BanglaBART Fine-tuned (Ours)',
                color=PALETTE['primary'], edgecolor='white')

    for bars in [b1, b2]:
        for bar in bars:
            h = bar.get_height()
            ax.text(bar.get_x() + bar.get_width()/2, h + 0.3,
                    f'{h:.1f}', ha='center', va='bottom', fontsize=10)

    ax.set_xticks(x)
    ax.set_xticklabels(['ROUGE-1','ROUGE-2','ROUGE-L'])
    ax.set_ylabel('Score (%)')
    ax.set_title('BanglaSum — Model Comparison')
    ax.legend()
    plt.tight_layout()
    save_fig('08_model_comparison')

    return comp_df


# ============================================================================
# 13. Human Evaluation Template
# ============================================================================
def export_human_eval_template(test_sample):
    # Export 20 random samples for human annotation
    human_eval_df = test_sample.sample(20, random_state=SEED)[[
        'text', 'summary', 'generated_summary'
    ]].copy().reset_index(drop=True)

    human_eval_df.columns = ['Article', 'Reference_Summary', 'Generated_Summary']

    # Add blank rating columns
    for criterion in ['Fluency_1_5', 'Adequacy_1_5', 'Conciseness_1_5', 'Faithfulness_1_5']:
        human_eval_df[criterion] = ''

    human_eval_path = RESULTS_DIR / 'human_evaluation_form.csv'
    human_eval_df.to_csv(human_eval_path, index=True, index_label='Sample_ID')
    print(f"✅ Human eval form saved: {human_eval_path}")
    print(f"   Samples: {len(human_eval_df)}")
    print(f"   Criteria: Fluency · Adequacy · Conciseness · Faithfulness (1–5 scale)")
    print(human_eval_df[['Reference_Summary','Generated_Summary']].head(3))

    return human_eval_df


# ============================================================================
# 14. Save All Results
# ============================================================================
def save_all_results(test_sample, rouge_scores, bert_f1_mean, N_TEST, comp_df, trainer):
    # ── 1. Full predictions CSV ───────────────────────────────────────────────
    pred_path = RESULTS_DIR / 'predictions.csv'
    test_sample[['text','summary','generated_summary',
                 'rouge1','rouge2','rougeL','bertscore_f1',
                 'gen_len','ref_len']].to_csv(pred_path, index=False)
    print(f"✅ Predictions saved : {pred_path}")

    # ── 2. Metrics JSON ───────────────────────────────────────────────────────
    final_metrics = {
        'model'       : MODEL_NAME,
        'n_test'      : N_TEST,
        'ROUGE-1'     : round(rouge_scores['ROUGE-1'], 4),
        'ROUGE-2'     : round(rouge_scores['ROUGE-2'], 4),
        'ROUGE-L'     : round(rouge_scores['ROUGE-L'], 4),
        'BERTScore_F1': round(bert_f1_mean, 4),
        'avg_gen_len' : round(test_sample['gen_len'].mean(), 2),
        'avg_ref_len' : round(test_sample['ref_len'].mean(), 2),
    }
    metrics_path = RESULTS_DIR / 'metrics.json'
    with open(metrics_path, 'w', encoding='utf-8') as f:
        json.dump(final_metrics, f, ensure_ascii=False, indent=2)
    print(f"✅ Metrics saved    : {metrics_path}")

    # ── 3. Comparison CSV ─────────────────────────────────────────────────────
    comp_path = RESULTS_DIR / 'model_comparison.csv'
    comp_df.to_csv(comp_path, index=False)
    print(f"✅ Comparison saved : {comp_path}")

    # ── 4. Training log JSON ──────────────────────────────────────────────────
    log_path = RESULTS_DIR / 'train_log.json'
    with open(log_path, 'w') as f:
        json.dump(trainer.state.log_history, f, indent=2)
    print(f"✅ Training log saved: {log_path}")

    # ── Summary printout ──────────────────────────────────────────────────────
    print("\n" + "═"*50)
    print("🏁  FINAL RESULTS — BanglaSum")
    print("═"*50)
    for k, v in final_metrics.items():
        print(f"  {k:<20}: {v}")
    print("═"*50)
    print(f"\n📁 All figures saved to : {FIGURES_DIR}")
    print(f"📁 All results saved to : {RESULTS_DIR}")
    print(f"📁 Best model saved to  : {OUTPUT_DIR / 'best_model'}")

    return final_metrics


# ============================================================================
# 15. File Summary
# ============================================================================
# banglasum_results/predictions.csv           -> All test predictions with per-sample scores
# banglasum_results/metrics.json              -> Final aggregated evaluation metrics
# banglasum_results/model_comparison.csv      -> Baseline vs BanglaBART comparison
# banglasum_results/train_log.json            -> Full training loss history
# banglasum_results/human_evaluation_form.csv -> Ready-to-fill annotator sheet
# banglasum_figures/01_dataset_statistics.png -> Dataset length distributions
# banglasum_figures/02_training_curves.png    -> Train/val loss curves
# banglasum_figures/03_metric_scores.png      -> Final metric bar chart
# banglasum_figures/04_rouge_distributions.png-> Per-sample ROUGE histograms
# banglasum_figures/05_rouge_vs_bertscore.png -> ROUGE vs BERTScore scatter
# banglasum_figures/06_length_analysis.png    -> Reference vs generated length
# banglasum_figures/07_metric_boxplots.png    -> Metric box plots
# banglasum_figures/08_model_comparison.png   -> Baseline vs BanglaBART bar chart
# banglasum_output/best_model/                -> Fine-tuned model weights + tokenizer


# ============================================================================
# MAIN PIPELINE
# ============================================================================
def main():
    # 4. Load & explore data
    raw, train_df, val_df, test_df = load_and_explore_dataset()

    # 5. Tokenizer & preprocessing
    tokenizer, tokenized = build_tokenizer_and_tokenize(raw)

    # 6. Model loading & fine-tuning
    model = load_model()
    trainer = build_trainer(model, tokenizer, tokenized)
    train_model(trainer, tokenizer)

    # 7. Training curves
    plot_training_curves(trainer)

    # 8. Generate summaries on test set
    test_sample, N_TEST = generate_on_test_set(model, tokenizer, test_df)

    # 9. Automatic evaluation metrics
    scorer, rouge_scores, r1_list, r2_list, rl_list = compute_rouge(test_sample)
    F1, bert_f1_mean = compute_bertscore(test_sample)
    metrics_df = summarize_metrics(test_sample, rouge_scores, bert_f1_mean,
                                    r1_list, r2_list, rl_list, F1)

    # 10. Evaluation visualizations
    plot_metric_scores(metrics_df)
    plot_rouge_distributions(test_sample)
    plot_rouge_vs_bertscore(test_sample)
    plot_length_analysis(test_sample)
    plot_metric_boxplots(test_sample)

    # 11. Qualitative analysis
    qualitative_analysis(test_sample)

    # 12. Baseline comparison
    comp_df = baseline_comparison(test_sample, scorer)

    # 13. Human evaluation template
    export_human_eval_template(test_sample)

    # 14. Save all results
    save_all_results(test_sample, rouge_scores, bert_f1_mean, N_TEST, comp_df, trainer)


if __name__ == "__main__":
    main()
```
